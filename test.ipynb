```python
import cvxpy as cp

# Define variables
x = cp.Variable()
y = cp.Variable()

# Define objective
objective = cp.Maximize(x + y)

# Define constraints
constraints = [
    x + y <= 1,
    x >= 0,
    y >= 0
]

# Solve the problem
problem = cp.Problem(objective, constraints)
result = problem.solve()

print("Problem status:", problem.status)
print("Optimal value:", result)
print("x =", x.value)
print("y =", y.value)
```
